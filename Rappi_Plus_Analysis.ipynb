### Tratamiento de Valores Nulos (`NaN`) en Columnas Numéricas

En el dataset `Orders` (24,990 registros), las columnas numéricas `cantidad`, `precio_unitario` y `monto_descuento` conservan **50 valores nulos (`NaN`)** (24,940 registros no nulos). Se decidió **no eliminar estas filas ni imputar valores artificiales** por las siguientes razones analíticas y de negocio:

* **Preservación de datos transaccionales válidos:** Cada una de estas 50 filas representa una transacción real que contiene información completa y valiosa en dimensiones clave como `id_pedido`, `id_usuario`, `fecha_hora_pedido`, `pais`, `dispositivo`, `fuente_referencia` y `monto_total`.
* **Evitar sesgos en la imputación:** Reemplazar estos nulos con métricas de tendencia central (como la media o la mediana) introduciría valores ficticios en el volumen de piezas o montos de descuento, distorsionando el cálculo real del comportamiento transaccional.
* **Compatibilidad con el modelado en Power BI / DAX:**
  * **Conteo de tráfico y transacciones:** Permitirá contabilizar el volumen total de órdenes y el comportamiento por canal de adquisición sin perder registros.
  * **Cálculos agregados (`SUM`, `AVERAGE`):** Las funciones del lenguaje DAX en Power BI ignoran automáticamente los valores nulos (`BLANK`) al ejecutar sumas o promedios, por lo que las métricas financieras sobre piezas o montos se mantendrán 100% precisas sin afectar los denominadores.

```python
# PARTE 1: Totales del funnel
# ======================
query_totals = """
SELECT
    nombre_evento,
    COUNT(DISTINCT id_usuario) AS total_eventos
FROM events
GROUP BY nombre_evento
ORDER BY 
    CASE nombre_evento
        WHEN 'first_visit' THEN 1
        WHEN 'select_item' THEN 2
        WHEN 'add_to_cart' THEN 3
        WHEN 'begin_checkout' THEN 4
        WHEN 'add_payment_info' THEN 5
        WHEN 'purchase' THEN 6
        ELSE 7
    END;
"""
totals = pd.read_sql(query_totals, con=engine)
totals
```

### 📌 Nota sobre la Secuencia del Embudo y Anomalía de Volumen

Al analizar la secuencia natural de navegación (**`first_visit` $\rightarrow$ `select_item` $\rightarrow$ `add_to_cart` $\rightarrow$ `begin_checkout` $\rightarrow$ `add_payment_info` $\rightarrow$ `purchase`**) mediante el conteo de **usuarios únicos** (`COUNT(DISTINCT id_usuario)`), se identifica el siguiente comportamiento:

 * **`select_item`**: 7,582 usuarios únicos
* **`add_to_cart`**: 7,634 usuarios únicos

**Hallazgo:**
 Existe un superávit de **52 usuarios únicos** (+0.69%) en la etapa de `add_to_cart` en comparación con `select_item`. 

 **Explicación técnica/negocio:**
 Este comportamiento no representa un error en el procesamiento de datos, sino una característica del flujo de la plataforma: existen usuarios que ejecutan la adición al carrito de forma directa (por ejemplo, a través de botones de compra rápida en listas generales, carruseles de recomendación o enlaces externos) sin necesidad de registrar la interacción previa de selección detallada (`select_item`).
 
 *Para preservar la transparencia del reporte comercial, los datos se presentan sin imputación ni exclusión arbitraria.*

## 3. Puntos Clave para la Entrega (Key Takeaways)

1. **Punto Principal de Pérdida de Clientes:** 
   El paso entre `begin_checkout` y `add_payment_info` representa el cuello de botella del embudo con una caída del **13.29%** (**1,042 usuarios perdidos**). Esto se debe a fricciones en la pasarela de pago (ausencia de pagos a un solo clic, costos/tarifas mostrados hasta el final o reingreso de datos de tarjeta).
2. **Eficiencia en la Conversión Final:**
   Una vez que el usuario ingresa su información de pago, el abandono es prácticamente nulo (**0.16%** / 10 usuarios), confirmando que la transacción final no tiene fallas técnicas.

```python
# PARTE 2: Conversiones
# ======================

query_conversion = """
WITH cte_first_visit AS (
    SELECT COUNT(DISTINCT(id_usuario)) AS total 
    FROM events 
    WHERE nombre_evento = 'first_visit'
),
cte_select_item AS (
    SELECT COUNT(DISTINCT(id_usuario)) AS total 
    FROM events 
    WHERE nombre_evento = 'select_item'
),
cte_add_to_cart AS (
    SELECT COUNT(DISTINCT(id_usuario)) AS total 
    FROM events 
    WHERE nombre_evento = 'add_to_cart'
),
cte_begin_checkout AS (
    SELECT COUNT(DISTINCT(id_usuario)) AS total 
    FROM events 
    WHERE nombre_evento = 'begin_checkout'
),
cte_add_payment_info AS (
    SELECT COUNT(DISTINCT(id_usuario)) AS total 
    FROM events 
    WHERE nombre_evento = 'add_payment_info'
),
cte_purchase AS (
    SELECT COUNT(DISTINCT(id_usuario)) AS total 
    FROM events 
    WHERE nombre_evento = 'purchase'
)
SELECT
    -- 1. Totales de cada evento
    (SELECT total FROM cte_first_visit) AS first_visit_total,
    (SELECT total FROM cte_select_item) AS select_item_total,
    (SELECT total FROM cte_add_to_cart) AS add_to_cart_total,
    (SELECT total FROM cte_begin_checkout) AS begin_checkout_total,
    (SELECT total FROM cte_add_payment_info) AS add_payment_info_total,
    (SELECT total FROM cte_purchase) AS purchase_total,

    -- 2. Cálculo de Drop-off % (Pérdida paso a paso)
    ROUND(
        ((SELECT total FROM cte_first_visit) - (SELECT total FROM cte_select_item)) * 100.0 / 
        NULLIF((SELECT total FROM cte_first_visit), 0), 2
    ) AS dropoff_first_to_select_pct,

    ROUND(
        ((SELECT total FROM cte_select_item) - (SELECT total FROM cte_add_to_cart)) * 100.0 / 
        NULLIF((SELECT total FROM cte_select_item), 0), 2
    ) AS dropoff_select_to_cart_pct,

    ROUND(
        ((SELECT total FROM cte_add_to_cart) - (SELECT total FROM cte_begin_checkout)) * 100.0 / 
        NULLIF((SELECT total FROM cte_add_to_cart), 0), 2
    ) AS dropoff_cart_to_checkout_pct,

    ROUND(
        ((SELECT total FROM cte_begin_checkout) - (SELECT total FROM cte_add_payment_info)) * 100.0 / 
        NULLIF((SELECT total FROM cte_begin_checkout), 0), 2
    ) AS dropoff_checkout_to_payment_pct,

    ROUND(
        ((SELECT total FROM cte_add_payment_info) - (SELECT total FROM cte_purchase)) * 100.0 / 
        NULLIF((SELECT total FROM cte_add_payment_info), 0), 2
    ) AS dropoff_payment_to_purchase_pct
;
"""


from sqlalchemy import text

conversion = pd.read_sql(text(query_conversion), con=engine)
conversion
```

**Validación de datos del funnel: Usuarios que hicieron 'add_to_cart' pero NUNCA registraron 'select_item'**

```python
query_usuarios_salto = """
WITH usuarios_cart AS (
    SELECT DISTINCT(id_usuario)
    FROM events
    WHERE nombre_evento = 'add_to_cart'
),
usuarios_select AS(
    SELECT DISTINCT(id_usuario)
    FROM events
    WHERE nombre_evento = 'select_item'
)
SELECT
    COUNT(c.id_usuario) AS usuarios_directos_carrito
FROM usuarios_cart c
LEFT JOIN usuarios_select s ON c.id_usuario = s.id_usuario
WHERE s.id_usuario IS NULL;
"""
usuarios_salto = pd.read_sql(text(query_usuarios_salto), con=engine)
usuarios_salto
```

**Validación de datos del funnel: Usuarios que hicieron 'select_item' pero NUNCA registraron 'add_to_cart'**

```python
# Usuarios que hicieron 'select_item' pero NUNCA llegaron a 'add_to_cart'
query_abandonos_select = """
WITH usuarios_select AS (
    SELECT DISTINCT id_usuario
    FROM events
    WHERE nombre_evento = 'select_item'
),
usuarios_cart AS (
    SELECT DISTINCT id_usuario
    FROM events
    WHERE nombre_evento = 'add_to_cart'
)
SELECT 
    COUNT(s.id_usuario) AS usuarios_que_abandonaron_en_select
FROM usuarios_select s
LEFT JOIN usuarios_cart c ON s.id_usuario = c.id_usuario
WHERE c.id_usuario IS NULL;
"""

abandonos = pd.read_sql(text(query_abandonos_select), con=engine)
abandonos
```

```python
# Retención por cohortes
# ======================
# NOTA: En la línea 10 se utiliza SUBSTRING(u.fecha_registro FROM 1 FOR 7) debido a que el formato de origen 
# en la base de datos para la columna fecha_registro es texto (VARCHAR), por lo que las transformaciones 
# a datetime en Pandas solo aplican en la memoria local del notebook.

query_cohort_retention_final = '''
WITH base_retencion AS (
    SELECT 
        SUBSTRING(u.fecha_registro FROM 1 FOR 7) AS cohorte_mes, 
        u.id_usuario,
        a.dias_despues_registro,
        a.activo
    FROM users u
    LEFT JOIN user_activity a ON u.id_usuario = a.id_usuario
)
SELECT 
    cohorte_mes,
    -- 1. Tamaño base de la cohorte (Semana 0)
    COUNT(DISTINCT id_usuario) AS tamano_cohorte,

    -- 2. Conteo de usuarios retenidos activos por semana
    COUNT(DISTINCT CASE WHEN dias_despues_registro = 7 AND activo = 1 THEN id_usuario END) AS retenido_w1,
    COUNT(DISTINCT CASE WHEN dias_despues_registro = 14 AND activo = 1 THEN id_usuario END) AS retenido_w2,
    COUNT(DISTINCT CASE WHEN dias_despues_registro = 21 AND activo = 1 THEN id_usuario END) AS retenido_w3,

    -- 3. Tasa de retención semanal (%)
    ROUND(
        COUNT(DISTINCT CASE WHEN dias_despues_registro = 7 AND activo = 1 THEN id_usuario END) * 100.0 / 
        NULLIF(COUNT(DISTINCT id_usuario), 0), 2
    ) AS semana_1,

    ROUND(
        COUNT(DISTINCT CASE WHEN dias_despues_registro = 14 AND activo = 1 THEN id_usuario END) * 100.0 / 
        NULLIF(COUNT(DISTINCT id_usuario), 0), 2
    ) AS semana_2,

    ROUND(
        COUNT(DISTINCT CASE WHEN dias_despues_registro = 21 AND activo = 1 THEN id_usuario END) * 100.0 / 
        NULLIF(COUNT(DISTINCT id_usuario), 0), 2
    ) AS semana_3

FROM base_retencion
GROUP BY cohorte_mes
ORDER BY cohorte_mes;
'''

# EJECUCIÓN CORREGIDA: Usando text() y la conexión explícita (conn)
with engine.connect() as conn:
    cohorte_final = pd.read_sql(text(query_cohort_retention_final), con=conn)

cohorte_final
```

📊 **Observación Analítica: Tasa de Retención Periódica**

 En cohortes como **2025-03** y **2025-04**, la tasa de retención presenta repuntes en la **Semana 2** (subiendo de 41.38% a 43.09% y de 42.34% a 43.40%, respectivamente).
 
 **Interpretación:** 
 La métrica empleada evalúa la retención periódica (actividad en ventanas discretas de 7, 14 y 21 días). Esto evidencia un fenómeno de **reactivación de usuarios**: clientes que no registraron actividad en la primera semana posterior a su registro, pero que volvieron a interactuar en las semanas 2 o 3 debido a ciclos habituales de recompra o estímulos comerciales.

# Test estadístico: ¿La modificación en la UI del checkout impacta la **tasa de conversión de compra**?.



# Hipótesis principal (Conversión Global)
**H₀ (Hipótesis nula):** No existe una diferencia significativa en la tasa de conversión entre la versión de control y la versión de tratamiento ($p_{\text{control}} = p_{\text{tratamiento}}$).

**H₁ (Hipótesis alternativa):** Existe una diferencia estadísticamente significativa en la tasa de conversión entre la versión de control y la versión de tratamiento ($p_{\text{control}} \neq p_{\text{tratamiento}}$).
    
# Hipótesis secundaria (duración de la sesión- paramétrica)
**H₀ (Hipótesis nula):** No existe una diferencia significativa en la tasa de conversión entre la versión de control y la versión de tratamiento ($p_{\text{control}} = p_{\text{tratamiento}}$).

**H₁ (Hipótesis alternativa):** Existe una diferencia estadísticamente significativa en la tasa de conversión entre la versión de control y la versión de tratamiento ($p_{\text{control}} \neq p_{\text{tratamiento}}$).
   
**Test estadístico:** hipotesis principal- Z-Test de Proporciones  | Hipotesis secundaria- t-Student

**Nivel de significancia alpha:** 0.05

```python
# Importar csv
experiment = pd.read_csv('https://practicum-content.s3.amazonaws.com/datasets/experiment_checkout_ui.csv')

experiment.head(15)
```

```python
# Conversión absoluta y tasa de conversión por variante (control vs tratamiento)
conversion_variante = experiment.groupby('variante').agg(
    total_usuarios=('id_usuario', 'count'),
    usuarios_convirtieron=('convirtio', 'sum'),
    tasa_conversion=('convirtio', 'mean'),
    duracion_promedio=('duracion_sesion', 'mean')
).reset_index()

conversion_variante['tasa_conversion'] = (conversion_variante['tasa_conversion'] * 100).round(2)
conversion_variante['duracion_promedio'] = conversion_variante['duracion_promedio'].round(2)

print("--- RESUMEN POR VARIANTE ---")
conversion_variante
```

```python
# Tabla de contingencia: Conversión por Variante y Dispositivo
contingencia_dispositivo = experiment.groupby(['dispositivo', 'variante']).agg(
    total_usuarios=('id_usuario', 'count'),
    usuarios_convirtieron=('convirtio', 'sum'),
    tasa_conversion=('convirtio', lambda x: (x.mean() * 100).round(2)),
    duracion_promedio=('duracion_sesion', lambda x: x.mean().round(2))
).reset_index()

contingencia_dispositivo
```

```python
# Tabla de contingencia: Conversión por Variante y País
contingencia_pais = experiment.groupby(['pais', 'variante']).agg(
    total_usuarios=('id_usuario', 'count'),
    usuarios_convirtieron=('convirtio', 'sum'),
    tasa_conversion=('convirtio', lambda x: (x.mean() * 100).round(2)),
    duracion_promedio=('duracion_sesion', lambda x: x.mean().round(2))
).reset_index()

contingencia_pais
```

```python
from statsmodels.stats.proportion import proportions_ztest
import scipy.stats as stats

# Extraer conteos de conversiones y totales
exitos = experiment.groupby('variante')['convirtio'].sum().loc[['tratamiento', 'control']].values
muestras = experiment.groupby('variante')['convirtio'].count().loc[['tratamiento', 'control']].values

# Ejecutar Z-test para proporciones
z_stat, p_value_conv = proportions_ztest(count=exitos, nobs=muestras, alternative='two-sided')

# Ejecutar t-Student para Duración de Sesión (Prueba Paramétrica)
duracion_control = experiment[experiment['variante'] == 'control']['duracion_sesion']
duracion_trat = experiment[experiment['variante'] == 'tratamiento']['duracion_sesion']
t_stat, p_value_dur = stats.ttest_ind(duracion_control, duracion_trat, equal_var=False)

# Imprimir Resultados
alpha = 0.05
print("=== RESULTADOS DE LA PRUEBA DE HIPÓTESIS ===")
print(f"Z-statistic (Conversión): {z_stat:.4f} | p-value: {p_value_conv:.4f}")
if p_value_conv < alpha:
    print("-> Rechazamos H0: La diferencia en conversión ES estadísticamente significativa.")
else:
    print("-> No se rechaza H0: No hay evidencia suficiente de cambio en la conversión.\n")

print(f"T-statistic (Duración): {t_stat:.4f} | p-value: {p_value_dur:.4f}")
if p_value_dur < alpha:
    print("-> Rechazamos H0: La diferencia en duración de sesión ES estadísticamente significativa.")
else:
    print("-> No se rechaza H0: La duración promedio entre grupos es estadísticamente equivalente.")
```

### 📊 Resultados de las Pruebas de Hipótesis (A/B Test)

Nivel de significancia establecido: $\alpha = 0.05$

* **1. Tasa de Conversión (Z-test de 2 proporciones)**

    **Hipótesis Nula (H0):** La tasa de conversión del grupo de tratamiento es igual a la del grupo de control.  
    **Hipótesis Alternativa (H1):** La tasa de conversión del grupo de tratamiento es diferente a la del grupo de control.

    **Estadístico Z: 0.8133 | p-value: 0.4161**

    **Resultado:** Como **p-value > 0.05**, no se rechaza la hipótesis nula **(H0)**. No existe evidencia estadística suficiente de una diferencia significativa en la tasa de conversión entre el grupo de control **(15.69%)** y tratamiento **(16.29%)**.

* **2. Duración de la Sesión (Prueba t de Student independiente)**

    **Hipótesis Nula (H0):** La duración promedio de la sesión en el grupo de tratamiento es igual a la del grupo de control.  
    **Hipótesis Alternativa (H1):** La duración promedio de la sesión en el grupo de tratamiento es diferente a la del grupo de control.

    **Estadístico t: 1.4454 | p-value: 0.1484**

    **Resultado:** Como **p-value > 0.05**, no se rechaza la hipótesis nula **(H0)**. La duración promedio de la sesión (**161.04 s** en control vs **158.70 s** en tratamiento) es estadísticamente equivalente entre ambos grupos.
